```python
# prompt: save print(response.taxt) in doc file and save

from google.colab import userdata
import google.generativeai as genai
import os

api = userdata.get('GEMINI_API_KEY')
if api:
  print("working")
else:
  print("not")

!pip install google.generativeai -q -U

genai.configure(api_key=api) # Use the API key from userdata

model_name = "gemini-1.5-flash"
model = genai.GenerativeModel(model_name)

# Create the output directory if it doesn't exist
output_dir = "/content/drive/MyDrive/gemini_outputs" # Or any other location on your drive
os.makedirs(output_dir, exist_ok=True)

responses = []

try:
  response = model.generate_content("Can LLM can OS?")
  responses.append(response.text)
  print(response.text)

  response = model.generate_content("terms/parameters to explains: Message, Model, Max Completion Tokens Stream, temperature, Top_p,tools")
  responses.append(response.text)
  print(response.text)

  # Save the responses to a .doc file (using a text file as .doc is not directly supported)
  filepath = os.path.join(output_dir, "gemini_output.txt")  # Change the filename as needed
  with open(filepath, "w") as f:
    for text in responses:
      f.write(text + "\n\n")

  print(f"Responses saved to {filepath}")

except Exception as e:
  print(f"An error occurred: {e}")
```
